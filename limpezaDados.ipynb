```python
# C:\Users\...\Documents\CSBC2026\CRIA\data\tab_cad_Dez2023.csv
# C:\Users\...\Documents\CSBC2026\CRIA\data\tab_cad_Dez2024.csv
# C:\Users\...\Documents\CSBC2026\CRIA\data\tab_cad_Junho2025.csv

"""
CadÚnico (2023, 2024, 2025) — Leitura robusta + anonimização (LGPD) + harmonização (UNIÃO DE COLUNAS) + exportação

O que este script faz (end-to-end):
1) Lê CSVs grandes (CadÚnico) de forma robusta:
   - detecta delimitador automaticamente (, ; | \t)
   - usa engine="python" (mais tolerante) e NÃO usa low_memory (incompatível)
   - ignora e registra linhas problemáticas (bad lines) em arquivo de log
   - tenta encoding utf-8 e faz fallback para latin1
2) Remove colunas sensíveis (PII / identificadores diretos / alto risco de reidentificação):
   - nomes, CPF, RG, NIS, apelido, título de eleitor, certidões, endereço detalhado,
     CEP, telefones, entrevistador etc.
3) Mantém:
   - (a) por padrão, mantém todas as colunas "não sensíveis" disponíveis em cada ano
   - (b) opcionalmente, você pode ativar KEEP_ONLY_MINIMAL_FOR_PAPER para reduzir para um conjunto mínimo
4) Harmoniza por UNIÃO de colunas (não interseção):
   - cria colunas ausentes em um ano com NA
   - alinha a ordem das colunas entre anos
   - adiciona coluna 'ano'
   - cria alvo binário inseg_alim_bin quando existir o indicador (2024+)
5) Exporta:
   - CSV anonimizado por ano
   - CSV consolidado por união de colunas (com NA nas colunas que não existiam no ano)
"""

from __future__ import annotations

import csv
from pathlib import Path
from typing import Dict, List, Optional, Iterable

import pandas as pd


# =========================================================
# CONFIG
# =========================================================
BASE_PATH = Path(r"C:\Users\...\Documents\CSBC2026\CRIA\data")
OUTPUT_PATH = BASE_PATH / "anon_outputs"
LOG_PATH = BASE_PATH / "logs"

OUTPUT_PATH.mkdir(parents=True, exist_ok=True)
LOG_PATH.mkdir(parents=True, exist_ok=True)

FILES: Dict[str, Path] = {
    "2023": BASE_PATH / "tab_cad_Dez2023.csv",
    "2024": BASE_PATH / "tab_cad_Dez2024.csv",
    "2025": BASE_PATH / "tab_cad_Junho2025.csv",
}

# Use chunks para bases muito grandes (recomendado).
CHUNKSIZE: Optional[int] = 250_000

# Se True: reduz para um conjunto mínimo "seguro" (recomendado se for compartilhar microdados).
# Se False: mantém todas as colunas não-sensíveis (mas ainda anonimiza).
KEEP_ONLY_MINIMAL_FOR_PAPER = False

# =========================================================
# LISTA DE EXCLUSÃO (PII / ALTO RISCO)
# =========================================================
SENSITIVE_COLUMNS = [
    # Identificadores diretos (pessoa)
    "p.nom_pessoa",
    "p.nom_apelido_pessoa",
    "p.num_cpf_pessoa",
    "p.num_identidade_pessoa",
    "p.num_nis_pessoa_atual",
    "p.nom_completo_mae_pessoa",
    "p.nom_completo_pai_pessoa",

    # Documentos/certidões/título eleitor/cartórios
    "p.cod_certidao_registrada_pessoa",
    "p.cod_certidao_civil_pessoa",
    "p.cod_livro_termo_certid_pessoa",
    "p.cod_folha_termo_certid_pessoa",
    "p.cod_termo_matricula_certid_pessoa",
    "p.cod_cartorio_certid_pessoa",
    "p.num_cart_trab_prev_soc_pessoa",
    "p.num_serie_trab_prev_soc_pessoa",
    "p.num_titulo_eleitor_pessoa",
    "p.num_zona_tit_eleitor_pessoa",
    "p.num_secao_tit_eleitor_pessoa",

    # Datas de emissão / órgão emissor
    "p.dta_emissao_ident_pessoa",
    "p.sig_orgao_emissor_pessoa",
    "p.sig_uf_ident_pessoa",
    "p.dta_emissao_cart_trab_pessoa",
    "p.sig_uf_cart_trab_pessoa",

    # Endereço detalhado (família)
    "d.nom_tip_logradouro_fam",
    "d.nom_titulo_logradouro_fam",
    "d.nom_logradouro_fam",
    "d.num_logradouro_fam",
    "d.des_complemento_fam",
    "d.des_complemento_adic_fam",
    "d.num_cep_logradouro_fam",
    "d.txt_referencia_local_fam",
    "d.nom_localidade_fam",

    # Contatos
    "d.num_ddd_contato_1_fam",
    "d.num_tel_contato_1_fam",
    "d.num_tel_contato_2_fam",
    "d.num_ddd_contato_2_fam",
    "d.ic_tipo_contato_1_fam",
    "d.ic_envo_sms_contato_1_fam",
    "d.ic_tipo_contato_2_fam",
    "d.ic_envo_sms_contato_2_fam",

    # Entrevistador (PII)
    "d.nom_entrevistador_fam",
    "d.num_cpf_entrevistador_fam",

    # Identificadores administrativos de alto risco para reidentificação externa
    "d.cod_familiar_fam",
    "p.cod_familiar_fam_origem",
    "d.cod_unidade_territorial_fam",
    "d.nom_unidade_territorial_fam",
    "d.ref_cad",
    "d.ref_pbf",
    "p.ref_cad",
    "p.ref_pbf",
]

# Conjunto mínimo de variáveis “seguras” (agregáveis) para paper
SAFE_KEEP_MINIMAL = [
    "d.cd_ibge",
    "d.vlr_renda_media_fam",
    "d.vlr_renda_total_fam",
    "d.fx_rfpc",
    "d.qtd_pessoas_domic_fam",
    "d.qtd_familias_domic_fam",
    "d.qtd_comodos_domic_fam",
    "d.qtd_comodos_dormitorio_fam",
    "d.val_desp_energia_fam",
    "d.val_desp_agua_esgoto_fam",
    "d.val_desp_gas_fam",
    "d.val_desp_alimentacao_fam",
    "d.val_desp_transpor_fam",
    "d.val_desp_aluguel_fam",
    "d.val_desp_medicamentos_fam",
    "d.cod_material_piso_fam",
    "d.cod_material_domic_fam",
    "d.cod_agua_canalizada_fam",
    "d.cod_abaste_agua_domic_fam",
    "d.cod_banheiro_domic_fam",
    "d.cod_escoa_sanitario_domic_fam",
    "d.cod_destino_lixo_domic_fam",
    "d.cod_iluminacao_domic_fam",
    "d.cod_calcamento_domic_fam",

    # marcadores sociais (podem não existir em todos os anos — serão criados como NA)
    "d.ind_risco_scl_inseg_alim",
    "d.ind_risco_scl_vlco_drts",
    "p.ind_identidade_genero",
    "p.ind_tipo_identidade_genero",
    "p.ind_transgenero",
    "d.marc_pbf",
    "p.marc_pbf",
    "p.marc_pab",
]

# =========================================================
# UTIL: detectar delimitador
# =========================================================
def sniff_delimiter(file_path: Path, n_lines: int = 20) -> str:
    """Detecta delimitador provável entre [',',';','\\t','|'] a partir de um sample."""
    with open(file_path, "r", encoding="utf-8", errors="ignore") as f:
        sample = "".join([f.readline() for _ in range(n_lines)])

    try:
        dialect = csv.Sniffer().sniff(sample, delimiters=[",", ";", "\t", "|"])
        return dialect.delimiter
    except Exception:
        # fallback simples: escolhe o que aparece mais
        return ";" if sample.count(";") >= sample.count(",") else ","

# =========================================================
# UTIL: iterador robusto por chunks (evita explodir RAM)
# =========================================================
def iter_csv_robust(file_path: Path, chunksize: int) -> Iterable[pd.DataFrame]:
    """
    Itera o CSV em chunks com engine='python' (tolerante),
    registrando linhas ruins e tentando utf-8 com fallback latin1.

    Importante:
    - low_memory NÃO é suportado com engine='python' => não usar.
    """
    sep = sniff_delimiter(file_path)
    bad_lines_log = LOG_PATH / f"bad_lines_{file_path.stem}.txt"

    def bad_line_handler(bad_line: List[str]):
        with open(bad_lines_log, "a", encoding="utf-8") as lf:
            lf.write("|".join(bad_line) + "\n")
        return None

    def _reader(enc: str):
        return pd.read_csv(
            file_path,
            sep=sep,
            engine="python",
            encoding=enc,
            encoding_errors="ignore",
            on_bad_lines=bad_line_handler,  # pandas >= 1.4
            quotechar='"',
            chunksize=chunksize,
        )

    try:
        yield from _reader("utf-8")
    except Exception:
        yield from _reader("latin1")


def read_csv_robust(file_path: Path, chunksize: Optional[int]) -> pd.DataFrame:
    """Lê o CSV inteiro. Se chunksize != None, concatena os chunks no final."""
    if chunksize is None:
        # leitura única (mais simples, exige RAM)
        sep = sniff_delimiter(file_path)
        return pd.read_csv(
            file_path,
            sep=sep,
            engine="python",
            encoding="utf-8",
            encoding_errors="ignore",
            on_bad_lines="skip",
            quotechar='"',
        )

    chunks = []
    for chunk in iter_csv_robust(file_path, chunksize=chunksize):
        chunks.append(chunk)
    return pd.concat(chunks, ignore_index=True)

# =========================================================
# ANONIMIZAÇÃO: remover colunas sensíveis + opção "mínimo"
# =========================================================
def anonymize_df(df: pd.DataFrame) -> pd.DataFrame:
    """
    Remove PII e, opcionalmente, reduz para colunas mínimas.
    """
    df2 = df.drop(columns=[c for c in SENSITIVE_COLUMNS if c in df.columns], errors="ignore")

    if KEEP_ONLY_MINIMAL_FOR_PAPER:
        keep_cols = [c for c in SAFE_KEEP_MINIMAL if c in df2.columns]
        df2 = df2[keep_cols].copy()

    return df2

# =========================================================
# NORMALIZAÇÃO + DERIVADAS
# =========================================================
def normalize_and_derive(df: pd.DataFrame, year: str) -> pd.DataFrame:
    """
    - Adiciona coluna ano
    - Normaliza colunas numéricas-chave (coerce)
    - Cria inseg_alim_bin se existir d.ind_risco_scl_inseg_alim
      (mapeamento: 1 -> 1 (em risco), 2 -> 0 (sem risco))
    """
    df = df.copy()
    df["ano"] = int(year)

    numeric_cols = [
        "d.vlr_renda_media_fam",
        "d.vlr_renda_total_fam",
        "d.qtd_pessoas_domic_fam",
        "d.val_desp_energia_fam",
        "d.val_desp_agua_esgoto_fam",
        "d.val_desp_gas_fam",
        "d.val_desp_alimentacao_fam",
    ]
    for c in numeric_cols:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")

    if "d.ind_risco_scl_inseg_alim" in df.columns:
        df["inseg_alim_bin"] = df["d.ind_risco_scl_inseg_alim"].map({1.0: 1.0, 2.0: 0.0})
    else:
        df["inseg_alim_bin"] = pd.NA

    return df

# =========================================================
# HARMONIZAÇÃO POR UNIÃO DE COLUNAS (NÃO INTERSEÇÃO)
# =========================================================
def harmonize_union(dfs: Dict[str, pd.DataFrame]) -> Dict[str, pd.DataFrame]:
    """
    Harmoniza por UNIÃO de colunas:
    - calcula o conjunto união de todas as colunas
    - para cada ano, cria colunas ausentes com NA
    - alinha a ordem de colunas igual para todos
    """
    all_cols = sorted(set().union(*(df.columns for df in dfs.values())))

    out: Dict[str, pd.DataFrame] = {}
    for year, df in dfs.items():
        df2 = df.copy()

        # cria colunas ausentes
        missing = [c for c in all_cols if c not in df2.columns]
        for c in missing:
            df2[c] = pd.NA

        # reordena colunas
        df2 = df2[all_cols].copy()
        out[year] = df2

    return out

# =========================================================
# MAIN
# =========================================================
def main():
    print("\n=== CadÚnico: leitura robusta + anonimização + harmonização (UNIÃO) + exportação ===")
    dfs: Dict[str, pd.DataFrame] = {}

    # 1) Ler + anonimizar + normalizar + exportar por ano
    for year, path in FILES.items():
        print("\n------------------------------")
        print(f"Processando {year}: {path}")

        df_raw = read_csv_robust(path, chunksize=CHUNKSIZE)
        print(f"  Registros lidos: {len(df_raw):,}")
        print(f"  Colunas originais: {len(df_raw.columns)}")

        df_anon = anonymize_df(df_raw)
        print(f"  Colunas após anonimização: {len(df_anon.columns)}")

        df_norm = normalize_and_derive(df_anon, year)

        out_year = OUTPUT_PATH / f"cadunico_{year}_anon.csv"
        df_norm.to_csv(out_year, index=False, encoding="utf-8")
        print(f"  Exportado: {out_year}")

        dfs[year] = df_norm

    # 2) Harmonização por UNIÃO
    dfs_union = harmonize_union(dfs)
    union_cols = list(dfs_union["2023"].columns)
    print("\n==============================")
    print(f"Total de colunas (UNIÃO 2023/2024/2025): {len(union_cols)}")

    # 3) Consolidado (concat com schema completo)
    df_all_union = pd.concat([dfs_union["2023"], dfs_union["2024"], dfs_union["2025"]], ignore_index=True)

    out_all = OUTPUT_PATH / "cadunico_2023_2025_union_anon.csv"
    df_all_union.to_csv(out_all, index=False, encoding="utf-8")

    print(f" Export consolidado (UNIÃO): {out_all}")
    print(" Pipeline finalizado.")

# if __name__ == "__main__":
main()


# Leitura + anonimização por ano
# 2023
# Registros: 2.013.017
# Colunas originais: 206
# Após anonimização: 156

# 2024
# Registros: 2.007.172
# Colunas originais: 211
# Após anonimização: 161

# 2025
# Registros: 2.018.516
# Colunas originais: 211
# Após anonimização: 161
```
